### Issues with the customer table
- custumer_name column has some trailing white spaces & some are captialized while some are lower case 
- customer_city column some entries are capitalized while some are lowecase and also leading and trailing white spaces
- signup_date has null values 


### Solution
- customer_name the trailing and leading white spaces would be removed and all entries would be converted to lowercase
- customers_city the trailing and leading white spaces would be removed and all entries would be converted to lowercase
- signup_date the right personnel would be consulted [The null values would be left as is]


```python
df = spark.table("ecommerce.bronze.customers")
display(df)
```

```python
from pyspark.sql.functions import lower, trim, col

df = df.withColumn('customer_name', lower(trim(col("customer_name"))))
display(df)
```

```python
df = df.withColumn('customer_city', lower(trim(col("customer_city"))))
display(df)
```

```python
df.write.format('Delta').mode('overwrite').saveAsTable('ecommerce.silver.customers')
display(spark.table("ecommerce.silver.customers").head(10))
```
