### Issues
- seller_city data format is inconsistent and to also has leading and trialing white spaces
### Solution
- seller_city would be transformed to lowercase and it would also be trimmed to remove white spaces 

```python
df = spark.table('ecommerce.bronze.sellers')
df.show()
```

```python
from pyspark.sql.functions import lower, trim, col

df = df.withColumn('seller_city', lower(trim(col("seller_city"))))

df.show()
```

```python
df.write.format('Delta').mode('overwrite').saveAsTable('ecommerce.silver.seller')
spark.table('ecommerce.silver.seller').show()
```
